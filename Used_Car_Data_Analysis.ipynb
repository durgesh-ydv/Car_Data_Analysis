```python
# share i20
share_i20 = df['car_name'].value_counts(normalize=True)*100
share_i20[0]
print(f"The avg share of {share_i20.head(1).index[0]} is {share_i20[0].round(2)}")
```

```python
# top 10 most sold car by brand
# Count number of cars by brand (car_name), normalize to get percentage share
brand_share = df['brand'].value_counts(normalize=True) * 100

# Display top 10 most sold car brands
top_10_brands = brand_share.head(10)

# Print nicely
for brand, share in top_10_brands.items():
    print(f"{brand}: {round(share, 2)}%")
```

```python
# avg selling price of the top selling brand in dataset
maruti = df[df['brand']=="Maruti"]
avg_selling_price_maruti=maruti['selling_price'].mean()
print(f"the avg selling price of maruti is {round(avg_selling_price_maruti, 2)}")
```

```python
# avg market share of the top 5 selling brand in dataset

# Step 1: Calculate market share (%) for all car brands
brand_share = df['car_name'].value_counts(normalize=True) * 100

# Step 2: Get the top 5 selling car brands
top_5_brands = brand_share.head(5)

# Step 3: Calculate average market share of these top 5 brands
avg_top5_share = top_5_brands.mean()

# Step 4: Print the result with brand names and shares
print("Top 5 Selling Car Brands and Their Market Shares:")
for brand, share in top_5_brands.items():
    print(f"- {brand}: {round(share, 2)}%")

print(f"\nThe average market share of the top 5 selling brands is: {round(avg_top5_share, 2)}%")
```

```python


# Step 1: Get top 10 costliest cars
top_10_costliest = df.sort_values(by='selling_price', ascending=False).head(10)

# Step 2: Plot using seaborn barplot
plt.figure(figsize=(20, 10))
sns.barplot(
    x='selling_price', 
    y='car_name', 
    data=top_10_costliest, 
    palette='magma'
)

# Add price labels next to bars
for index, value in enumerate(top_10_costliest['selling_price']):
    plt.text(value + 10000, index, f"₹{int(value)}", va='center')

plt.title('Top 10 Costliest Cars in Dataset')
plt.xlabel('Selling Price (₹)')
plt.ylabel('Car Name')
plt.tight_layout()
plt.show()
```
